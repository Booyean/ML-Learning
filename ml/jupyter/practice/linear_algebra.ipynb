```python
A_sum_axis1 = A.sum(axis=1, keepdims=True)
#行求和
A, A_sum_axis1, A_sum_axis1.shape
```

```python
A_sum_axis2 = A.sum(axis=2, keepdims=True)
#列求和
A, A_sum_axis2, A_sum_axis2.shape
```

```python
A_sum_axis0_1 = A.sum(axis=[0, 1])
A, A.shape, A_sum_axis0_1, A_sum_axis0_1.shape
```

```python
A_sum_axis0_1_2 = A.sum([0, 1, 2])
A, A.shape, A_sum_axis0_1_2, A_sum_axis0_1_2.shape, A.sum(), A.sum().shape
#标量的没有具体的size,所以[]内部没有具体的表示
```

如果我们不想舍去求和的维度，那么可以借助一下广播机制

```python
sum_A = A.sum(axis=2, keepdims=True)
sum_A
```

```python
#我们来看看没有keepdims=True会怎么样
sum_A = A.sum(axis=2)
sum_A,sum_A.shape
```

我们发现自然而然的就少了一个维度了，那么我们这样是无法进行广播操作的

```python
sum_A = A.sum(axis=2, keepdims=True)
sum_A
```

```python
A / sum_A, A
```
