## 数据处理

```python
import os
#os模块提供与操作系统相互交互的功能

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
os.makedirs(os.path.join('..','comfirm_example'), exist_ok=True)
#使用该内函数构造路径'../data'，表示当前目录的上一级目录下的data文件夹
#外函数创建该路径下的目录，如果目录已存在且exist_ok=True,就不报错
#这一步是在创建文件夹
data_file = os.path.join('..', 'data', 'house_tiny.csv')
ce = os.path.join('..', 'comfirm_example', 'test.csv')#实际这一步我们已经创建好了文件，但是因为还没写，就是空文件，不显示
#构造路径'../data/house_tiny.csv',并赋值给该变量
with open(data_file, 'w') as f:
    #以写模式'w'打开文件data_file文件，并使用with语句保证文件在操作后可以顺利关闭
    f.write('NumRooms,Alley,Price\n')
    #给文件写入表头
    f.write('NA,Pave,127500\n')
    #向文件写入一行数据
    f.write('2,NA,106000\n')
    f.write('3,NA,178100\n')
    f.write('NA,NA,140000\n')
with open(ce, 'w') as c:
    c.write('i, you, he\n')
    c.write('ZMY, LJX, Nobody\n')
```

```python
import pandas as pd
#pandas是数据处理和分析的库


data = pd.read_csv(data_file)
name = pd.read_csv(ce)
data
```

知识解释：
1. csv文件是常见的数据储存格式，用于储存***表格数据*** ，每一行代表一条记录，字段之间用逗号隔开
2. 'w'是文件打开方式的一种，即写模式，如果文件已经存在'w'模式会清除文件内容并从头开始写；如果文件不存在，则会创造新的文件
3. 一层目录可以理解为一层文件夹
4. 我们通过这样程序在系统的jupyter文件夹里创建了一个data文件夹，并在这个文件夹里创建以一个csv文件

缺失值处理：

```python
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]
#就是把表格中第一列和第二列的元素全部提取出来归为一个大类，特征；第三列提取出来归为标签（输出）
#其中的iloc函数是列表索引函数
inputs = inputs.fillna(inputs.mean())
print(inputs)
```

```python
#显然这种方法对string是不起作用的，用以下的归类方法
inputs = pd.get_dummies(inputs, dummy_na=True)
inputs
```

解释：
1. 通过pd.get_dummies()函数进行字符串的独热编码，我们通过Ture告知程序我们需要对NaN缺失值单独生成一个哑变量列
2. 通过引入独热编码函数get_dummies进行独热编码，独热编码函数的参数有多种，dummy_na是其中一种（默认是Fales）,其他的我暂时没有了解了
